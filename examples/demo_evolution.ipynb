# Trail demo — watching a cell evolve

A tiny numpy-only stand-in for a lecture: linear regression by gradient descent,
with a learning rate that's set far too high.

Run it top to bottom. The `train` cell will blow up. Then **fix it and re-run** —
that's exactly the history Trail exists to keep. Afterwards, on the Mac:

    trail log demo-evolution

```python
# trail: skip
import trail
trail.start("demo-evolution")
```

```python
# trail: cell data
import numpy as np

rng = np.random.default_rng(1337)
n_samples, n_features = 512, 8
X = rng.normal(size=(n_samples, n_features))
true_w = rng.normal(size=n_features)
y = X @ true_w + rng.normal(scale=0.5, size=n_samples)
print("X", X.shape, "y", y.shape)
```

```python
# trail: cell hparams
lr = 1.5
steps = 60
```

```python
# trail: cell train
# trail: cp loss explodes to inf — lr is way too high
w = np.zeros(n_features)

for step in range(steps):
    grad = 2 * X.T @ (X @ w - y) / n_samples
    w -= lr * grad
    if step % 20 == 0:
        print(f"{step:3d}/{steps}: {np.mean((X @ w - y) ** 2):.4f}")

loss = float(np.mean((X @ w - y) ** 2))
print("final loss", loss)
trail.metric("train_loss", loss)
```

### Now fix it

That loss went to infinity: each step overshoots the minimum by more than it started
from, so the updates grow instead of shrinking.

Go back to the **`hparams`** cell, set `lr = 0.5`, and re-run both cells. Then change
the `@cp` note on `train` to say what you fixed, e.g.

    # trail: cp dropped lr 1.5 → 0.5, loss converges now

Trail keeps both versions, folds the failed attempts in, and can explain the difference.

```python
# trail: cell plot
%matplotlib inline
import matplotlib.pyplot as plt

plt.figure(figsize=(5, 3))
plt.scatter(X @ true_w, X @ w, s=6, alpha=0.4)
plt.xlabel("true")
plt.ylabel("predicted")
plt.title(f"loss = {loss:.4f}")
plt.tight_layout()
plt.show()
```

```python
# trail: skip
trail.status(verbose=True)
trail.report()
```
